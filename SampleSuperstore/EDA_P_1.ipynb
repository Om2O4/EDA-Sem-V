#Importing Dataset and the libraries needed!

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv("/SampleSuperstore.csv")
```

#Data Analysis and grain identification -->30 grains


Sales and Revenue

```python
#Total overall sales across all records.
total_sales=df['Sales'].sum()
print(total_sales)
```

```python
#Average sales value per transaction.
avg_sales=df['Sales'].mean()
print(avg_sales)
```

```python
#Maximum and minimum sales value in a single transaction.
max_sales=df['Sales'].max()
min_sales=df['Sales'].min()
print(max_sales)
print(min_sales)
```

```python
#Total sales grouped by Category.
sales_by_category=df.groupby('Category')['Sales'].sum()
print(sales_by_category)
```

```python
#Total sales grouped by Sub-Category.
sales_by_subcategory=df.groupby('Sub-Category')['Sales'].sum()
print(sales_by_subcategory)
```

```python
#Average sales per Sub-Category.
avg_sales_by_subcategory=df.groupby('Sub-Category')['Sales'].mean()
print(avg_sales_by_subcategory)
```

```python
#Overall sales distribution across the four Regions.
sales_by_region=df.groupby('Region')['Sales'].sum()
print(sales_by_region)
```

```python
#Percentage contribution of each Segment to total sales.
total_sales=df['Sales'].sum()
sales_by_segment=df.groupby('Segment')['Sales'].sum()
segment_sales_percentage=(sales_by_segment/total_sales)*100
print(segment_sales_percentage)
```

Profitability & Discounts

```python
#Total overall profit generated.
total_profit=df['Profit'].sum()
print(total_profit)
```

```python
#Average profit per transaction.
avg_profit=df['Profit'].mean()
print(avg_profit)
```

```python
#Overall Profit Margin (Total Profit / Total Sales).
profit_margin=(total_profit/total_sales)*100
print(profit_margin)
```

```python
#Profit margin broken down by Category.
profit_margin_by_category=(df.groupby('Category')['Profit'].sum()/df.groupby('Category')['Sales'].sum())*100
print(profit_margin_by_category)
```

```python
#Identification of Sub-Categories operating at a net loss.
subcat_profit = df.groupby('Sub-Category')['Profit'].sum()
loss_making = subcat_profit[subcat_profit < 0]
print(loss_making.to_string())
```

```python
#The most profitable State.
state_profit = df.groupby('State')['Profit'].sum()
print(state_profit.idxmax(),state_profit.max())
```

```python
#The least profitable State (highest losses).
print(state_profit.idxmin(), state_profit.min())
```

```python
#Correlation coefficient between Discount and Profit
correlation_discount_profit=df['Discount'].corr(df['Profit'])
print(correlation_discount_profit)
```

```python
#Average profit segmented by Ship Mode
profit_by_shipmode=df.groupby('Ship Mode')['Profit'].mean()
print(profit_by_shipmode)
```

```python
#Impact of high discounts (> 20%) on profit
high_discount_profit=df[df['Discount']>0.2]['Profit'].mean()
print(high_discount_profit)
```

GEOGRAPHICAL INSIGHTS

```python
#Total number of unique cities
unique_cities=df['City'].nunique()
print(unique_cities)
```

```python
#Top 5 cities with the highest total sales
top_cities_sales=df.groupby('City')['Sales'].sum().nlargest(5)
print(top_cities_sales)
```

```python
#Bottom 5 cities with the lowest total profit
bottom_cities_profit=df.groupby('City')['Profit'].sum().nsmallest(5)
print(bottom_cities_profit)
```

```python
#Direct comparison of Sales vs. Profit by Region
sales_profit_by_region=df.groupby('Region')[['Sales','Profit']].sum()
print(sales_profit_by_region)
```

```python
#States with the highest total Quantity of items sold (Top 5)
state_quantity=df.groupby('State')['Quantity'].sum().nlargest(5)
print(state_quantity)
```

```python
#Average discount percentage given in each Region
avg_discount_by_region=df.groupby('Region')['Discount'].mean()
print(avg_discount_by_region)
```

```python
#Regional preferences for specific Categories
category_preferences=df.groupby(['Region','Category'])['Sales'].sum()
print(category_preferences)
```

Product and Inventory

```python
#Total absolute quantity of items sold
total_quantity=df['Quantity'].sum()
print(total_quantity)
```

```python
#Most frequently purchased Sub-Category (by transaction count)
most_purchased_subcategory=df['Sub-Category'].value_counts().idxmax()
print(most_purchased_subcategory)
```

```python
#Least frequently purchased Sub-Category (by transaction count)
least_purchased_subcategory=df['Sub-Category'].value_counts().idxmin()
print(least_purchased_subcategory)
```

```python
#Average quantity of items purchased per segment(transaction row)
avg_quantity_per_segment=df.groupby('Segment')['Quantity'].sum().mean()
print(avg_quantity_per_segment)
```

```python
#The Category that receives the highest average discount
avg_discount_by_category=df.groupby('Category')['Discount'].mean()
print(avg_discount_by_category.idxmax())
```

```python
#Statistical variance and standard deviation of sales across Sub-Categories
sales_variance=df.groupby('Sub-Category')['Sales'].var()
sales_std_dev=df.groupby('Sub-Category')['Sales'].std()
print(sales_variance)
print(sales_std_dev)
```

```python
#Total number of unique Sub-Categories
unique_subcategories=df['Sub-Category'].nunique()
print(unique_subcategories)
```

```python
#Average profit per single unit sold (Profit / Quantity)
avg_profit_per_unit=df['Profit'].sum()/df['Quantity'].sum()
print(avg_profit_per_unit)
```
