#DAY 2 - Scaling and Normilization

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from mlxtend.preprocessing import minmax_scaling

ks_17=pd.read_csv('/content/ks-projects-201801.csv')
np.random.seed(0)
```

```python
usd_goal = ks_17[['usd_goal_real']]
scaled_data = minmax_scaling(usd_goal, columns=['usd_goal_real'])

# plot the original & scaled data together to compare
fig, ax = plt.subplots(1, 2, figsize=(15, 3))
sns.distplot(ks_17.usd_goal_real, ax=ax[0])
ax[0].set_title("Original Data")
sns.distplot(scaled_data, ax=ax[1])
ax[1].set_title("Scaled data")
```

```python
goal_data = ks_17[['goal']]
scaled_goal_data = minmax_scaling(goal_data, columns=['goal'])
fig, ax = plt.subplots(1, 2, figsize=(15, 3))
sns.histplot(ks_17.goal, ax=ax[0], bins=50)
ax[0].set_title("Original Data")
sns.histplot(scaled_goal_data, ax=ax[1], bins=50, legend=False)
ax[1].set_title("Scaled data")
```

```python
index_of_positive_pledges = ks_17.usd_pledged_real > 0
positive_pledges = ks_17.usd_pledged_real.loc[index_of_positive_pledges]
normalized_pledges = stats.boxcox(positive_pledges)[0]

# plot both together to compare
fig, ax=plt.subplots(1,2)
sns.distplot(positive_pledges, ax=ax[0])
ax[0].set_title("Original Data")
sns.distplot(normalized_pledges, ax=ax[1])
ax[1].set_title("Normalized data")
```

```python
index_positive_pledges = ks_17.pledged > 0
positive_pledges = ks_17.pledged.loc[index_positive_pledges]

normalized_pledges = pd.Series(stats.boxcox(positive_pledges)[0],name='pledged', index=positive_pledges.index)
fig, ax = plt.subplots(1, 2, figsize=(15, 3))

#plot
sns.histplot(positive_pledges, ax=ax[0], bins=50)
ax[0].set_title("Original Data")
sns.histplot(normalized_pledges, ax=ax[1], bins=50)
ax[1].set_title("Normalized data")
```

#DAY 3 - Parsing Dates

```python
#import libs
import datetime
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

earth = pd.read_csv('/content/earthquake_database.csv')
vol = pd.read_csv('/content/vol_database.csv')
land = pd.read_csv('/content/landslide_catalog.csv')
```

```python
# print the first few rows of the date column
print(land['date'].head())
```

```python
land['date'].dtype
```

```python
# Check the data type of the Date column
print(earth['Date'].dtype)
earth['Date'].head()
```

```python
land['date_parsed'] = pd.to_datetime(land['date'], format = "%m/%d/%y")
print(land.head(10))
```

```python
land['date_parsed'].head()
```

```python
earth['date_parsed'] = pd.to_datetime(earth['Date'], format='mixed')
print(earth['date_parsed'].dtype)
```

```python
# 1. Force the conversion to a true datetime object (turns the 3 weirdly formatted rows into NaT)
earth['date_parsed'] = pd.to_datetime(earth['Date'], errors='coerce')

# 2. NOW you can successfully use the .dt accessor!
day_of_month_earthquakes = earth['date_parsed'].dt.day

# 3. Take a look at the first few rows to verify it worked
day_of_month_earthquakes.head()
```

```python
day_of_month_landslides = land['date_parsed'].dt.day
```

```python
# remove na's
day_of_month_landslides = day_of_month_landslides.dropna()

# plot the day of the month
sns.distplot(day_of_month_landslides, kde=False, bins=31)
```

```python
# 1. Drop the missing values (NaT) from our day of the month series
day_of_month_earthquakes = day_of_month_earthquakes.dropna()

# 2. Plot the distribution using 31 bins (one for each possible day of the month)
plt.figure(figsize=(10, 5))
sns.histplot(day_of_month_earthquakes, bins=31, kde=False)

# 3. Add some labels for clarity
plt.title("Earthquakes by Day of the Month")
plt.xlabel("Day of the Month")
plt.ylabel("Number of Earthquakes")
plt.show()
```
